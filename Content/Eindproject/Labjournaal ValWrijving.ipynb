```python
#Data processing and analysis:
L_naald = 0.0105 #Daadwerkelijke lengte van naald (m)

#Diameters zijn berekend met behulp van Fiji-software en schaal van L_naald
#Aangenomen dat alle druppels perfect bolvormig zijn, ook al zijn ze dat in de fotos niet helemaal
#Uit de fotos afgeleide diameters (m)
D_druppel1 = 0.003875 
D_druppel2 = 0.003993 
D_druppel3 = 0.003888

#Straal van druppels (m)
R_druppel1 = D_druppel1/2
R_druppel2 = D_druppel2/2
R_druppel3 = D_druppel3/2

#Volume van druppels (m^3)
#Aangenomen dat druppels perfect bolvormig zijn
V_druppel1 = (4/3)*np.pi*R_druppel1
V_druppel2 = (4/3)*np.pi*R_druppel2
V_druppel3 = (4/3)*np.pi*R_druppel3

#Massa van druppels (kg)
m_druppel1 = d_water*V_druppel1
m_druppel2 = d_water*V_druppel2
m_druppel3 = d_water*V_druppel3
```

**Describing the pattern in the processed data:**  


Integreren van Vgl. (14.7) geeft, onder verwaarlozing van de tweede orde correctie:

(14.8)
 $$
    s(t) = s_0 + v_0t + \frac{1}{2}gt^2 - \frac{\beta g^2}{12m}t^4
 $$
waarin 
 de afgelegde weg is bij 
. Als we een situatie kunnen creëren waarin bij 
 zowel de afgelegde weg als de snelheid nul zijn, dan volgt uit Vgl. (14.8):

(14.9)
 $$
   s(t) - \frac{1}{2}gt^2 = -\frac{\beta g^2}{12m}t^4
$$
 
 
In woorden staat hier dat de valafstand in lucht op ieder moment is gereduceerd t.o.v. de valafstand bij een vrije val met een hoeveelheid die evenredig is met de valtijd tot de vierde macht. De evenredigheidsconstante bevat de parameter 
, die zelf weer evenredig is met 
. Aan de hand van deze vergelijking kan een experiment worden bedacht waarbij bijvoorbeeld 
 voor een vallende waterdruppel wordt bepaald. In Figuur 14.8 is een voorbeeld van een opstelling weergeven die hiervoor gebruikt kan worden (en bij het practicum aanwezig is).

```python
#Calculations of e.a. measurement uncertainties, and providing final answers.
#Eventueel waardes van druppel1 gebruikt
def eom(beta,t):
    return ((0.5 * g * t**2) - (((beta * g**2) / (12 * m_druppel1)) * t**4))

popt, pcov = curve_fit(eom, Val_tijden, Hoogte_Data)
beta_fit = popt[0]
u_beta_fit = np.sqrt(np.diag(pcov))[0]

#Plot met fit
t_fit = np.linspace(min(Val_tijden), max(Val_tijden),100)
H_fit = eom(beta_fit, t_fit)

plt.figure()
plt.plot(Val_tijden, Hoogte_Data, 'r.', label='Data')
plt.plot(t_fit, H_fit, 'b--', label='Fit')
plt.xlabel('Valtijd (s)')
plt.ylabel('Valhoogte (m)')
plt.show()
```
